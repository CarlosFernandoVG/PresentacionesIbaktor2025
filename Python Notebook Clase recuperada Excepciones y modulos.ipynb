El tratamiendo de excepciones, al igual que otro bloque de código, puede estar dentro de otra para un tratamiento adecuado

```python
print(40)
try: 
    numerador = int(input("Escribe el numerado"))
    denominador = int(input("Escribe el denominador"))
    resultado = numerador / denominador
    print(resultado)
except ZeroDivisionError: 
    print("Oye, por que divides entre cero?") 
    print("Ingresa otro número DISTINTO DE CERO") 
    try: 
        denominador = int(input("Escribe el denominador"))
        resultado = numerador / denominador
        print(resultado)
    except ZeroDivisionError: 
        print("Con esta gente no se puede") 
print(50)
```

Podemos tratar diferentes tipos de excepciones

```python
print(40)
try: 
    numerador = int(input("Escribe el numerado"))
    denominador = int(input("Escribe el denominador"))
    resultado = numerador / denominador
    print(resultado)
except ZeroDivisionError: 
    print("Oye, por que divides entre cero?, :p") 
    print("Ingresa otro número DISTINTO DE CERO") 
    try: 
        denominador = int(input("Escribe el denominador"))
        resultado = numerador / denominador
        print(resultado)
    except ZeroDivisionError: 
        print("Con esta gente no se puede") 
except ValueError:
    try:
        print("Ingresaste un formato no válido, por favor ingresa un valor entero postivo")
        denominador = int(input("Escribe el denominador"))
        resultado = numerador / denominador
        print(resultado)
    except Exception:
        print("Error desconocido")
#La super clase de Excepciones(los objetos o sub clases) es Exception
#La podemos usar para tratar cualquier otro problema
except Exception:
    print("Error desconocido")
print(50)
```

El uso final de Exception garantizar que bajo cualquier problema tenemos cubierto el tema

```python
print(40)
conteo_errores = 0
try: 
    numerador = int(input("Escribe el numerado"))
    denominador = int(input("Escribe el denominador"))
    resultado = numerador / denominador
    print(resultado)
except ZeroDivisionError: 
    conteo_errores += 1
    print("Oye, por que divides entre cero?") 
    print("Ingresa otro número DISTINTO DE CERO") 
    try: 
        denominador = int(input("Escribe el denominador"))
        resultado = numerador / denominador
        print(resultado)
    except ZeroDivisionError: 
        print("Con esta gente no se puede") 
except ValueError:
    conteo_errores += 1
    print("Ingresaste un formato no válido, por favor ingresa un valor entero postivo")
    denominador = int(input("Escribe el denominador"))
    resultado = numerador / denominador
    print(resultado)
except Exception:
    conteo_errores += 1
    print("Error desconocido")
print(50)
print(conteo_errores)
```

Además de tratar todos los problemas, usamos finally

```python
print(40)
conteo_errores = 0
try: 
    numerador = int(input("Escribe el numerado"))
    denominador = int(input("Escribe el denominador"))
    resultado = numerador / denominador
    print(resultado)
except ZeroDivisionError: 
    conteo_errores += 1
    print("Oye, por que divides entre cero?") 
    print("Ingresa otro número DISTINTO DE CERO") 
    try: 
        denominador = int(input("Escribe el denominador"))
        resultado = numerador / denominador
        print(resultado)
    except ZeroDivisionError: 
        print("Con esta gente no se puede") 
except ValueError:
    conteo_errores += 1
    print("Ingresaste un formato no válido, por favor ingresa un valor entero postivo")
    denominador = int(input("Escribe el denominador"))
    resultado = numerador / denominador
    print(resultado)
#La super clase de Excepciones(los objetos o sub clases) es Exception
#La podemos usar para tratar cualquier otro problema
except Exception:
    conteo_errores += 1
    print("Error desconocido")
finally:
    print("Se acabo el juego")
print(50)
print(conteo_errores)
```

# Numpy

NumPy (Numerical Python) es una biblioteca fundamental para la computación científica con Python. 
Proporciona:
+ Arrays multidimensionales (más eficientes que las listas nativas)
+ Operaciones vectorizadas
+ Funciones matemáticas, estadísticas y de álgebra lineal

```python
import numpy as np
```

## Creación de Arrays

```python
# Desde listas de Python
a = np.array([1, 2, 3])
print("Array a:", a)
```

```python
type(a)
```

```python
# Rango de valores
b = np.arange(0, 10, 2)
print("Array b:", b)
```

```python
# Espaciado uniforme
c = np.linspace(0, 1, 5)
print("Array c:", c)
```

```python
# Matriz de ceros o unos
d = np.zeros((2, 3))
e = np.ones((3, 2))

print("Zeros:\n", d)
print("Ones:\n", e)
```

## Indexación y Slicing

```python
arr = np.array([10, 20, 30, 40, 50])
print("Elemento en posición 1:", arr[1])
print("Últimos dos elementos:", arr[-2:])
```

```python
matriz = np.array([[1, 2, 3], [4, 5, 6]])
print("Elemento (1,2):", matriz[1, 2])
print("Primera fila:", matriz[0])
print("Segunda columna:", matriz[:, 1])
```

## Operaciones Aritméticas y Vectorización

```python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print("Suma:", x + y)
print("Producto:", x * y)
print("Cuadrado:", x ** 2)
print("Cuadrado:", x ** y)
print("Seno:", np.sin(x))
```

```python
3*3*3*3*3*3
```

## Estadísticas Básicas

```python
data = np.array([10, 20, 20, 30, 40])

print("Media:", np.mean(data))
print("Mediana:", np.median(data))
print("Desviación estándar:", np.std(data))
print("Máximo:", np.max(data))
print("Índice del valor mínimo:", np.argmin(data))
```

## Broadcasting

Permite operar arrays de diferentes formas/tamaños sin necesidad de bucles.

```python
A = np.array([[1], [2], [3]])
B = np.array([10, 20])

# A se "extiende" para emparejarse con B
print("Resultado de A + B:\n", A + B)
```

## Álgebra Lineal

```python
mat = np.array([[1, 2], [3, 4]])
print("Transpuesta:\n", mat.T)
```

```python
# Inversa
inv = np.linalg.inv(mat)
print("Inversa:\n", inv)
```

```python
# Producto matricial
prod = np.dot(mat, inv)
print("Producto mat x inv:\n", prod)
```

## Comparación de rendimiento

```python
import time

lista = list(range(1_000_000))
arr = np.array(lista)

start = time.time()
[2*x for x in lista]
print("Lista:", time.time() - start)

start = time.time()
2 * arr
print("NumPy:", time.time() - start)
```

## Agrupamiento (groupby)

```python
data = {
    'Departamento': ['Ventas', 'Ventas', 'TI', 'TI', 'TI'],
    'Empleado': ['Ana', 'Luis', 'Carlos', 'Sofía', 'Pedro'],
    'Salario': [50000, 52000, 60000, 61000, 58000]
}
df3 = pd.DataFrame(data)

# Agrupar y resumir
print(df3.groupby('Departamento')['Salario'].mean())
```

## Unión y combinación

```python
df_a = pd.DataFrame({'ID': [1, 2], 'Nombre': ['Ana', 'Luis']})
df_b = pd.DataFrame({'ID': [1, 2], 'Edad': [23, 25]})

# Merge
df_merge = pd.merge(df_a, df_b, on='ID')
print(df_merge)

# Concatenar
df_concat = pd.concat([df_a, df_b], axis=1)
print(df_concat)
```

# Matplotlib

matplotlib es una biblioteca de visualización 2D de gráficos en Python. Se utiliza para generar gráficos de líneas, barras, histogramas, diagramas de dispersión, etc.

```python
import matplotlib.pyplot as plt
```

## Gráfico de línea simple

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Gráfico de línea")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.grid(True)
plt.show()
```

## Personalización de estilo

```python
plt.plot(x, y, color='red', linestyle='--', linewidth=2, marker='o', label='seno')
plt.title("Gráfica de seno con estilo")
plt.xlabel("Ángulo (rad)")
plt.ylabel("Valor")
plt.legend()
plt.show()
```

## Múltiples líneas en un solo gráfico

```python
y2 = np.cos(x)

plt.plot(x, y, label='seno')
plt.plot(x, y2, label='coseno')
plt.title("Seno vs Coseno")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

## Subplots (varios gráficos)

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 6))

axs[0, 0].plot(x, y)
axs[0, 0].set_title("Seno")

axs[0, 1].plot(x, y2)
axs[0, 1].set_title("Coseno")

axs[1, 0].plot(x, y + y2)
axs[1, 0].set_title("Seno + Coseno")

axs[1, 1].plot(x, y - y2)
axs[1, 1].set_title("Seno - Coseno")

plt.tight_layout()
plt.show()
```

## Gráficos de barras

```python
categorias = ['A', 'B', 'C', 'D']
valores = [23, 45, 56, 78]

plt.bar(categorias, valores, color='skyblue')
plt.title("Gráfico de barras")
plt.ylabel("Valores")
plt.show()
```

```python
plt.barh(categorias, valores, color='skyblue')
plt.title("Barras horizontales")
plt.show()
```

## Histogramas

```python
data = np.random.randn(1000)

plt.hist(data, bins=30, color='salmon', edgecolor='black')
plt.title("Histograma")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.show()
```

## Diagramas de dispersión

```python
x = np.random.rand(100)
y = np.random.rand(100)

plt.scatter(x, y, color='purple', alpha=0.6)
plt.title("Gráfico de dispersión")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
