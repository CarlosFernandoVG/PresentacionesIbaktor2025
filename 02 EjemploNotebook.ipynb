````python
def Adivina_n (limit1:int , limit2: int, n:int = 10) -> int:
    """Juego de adivinar un número

    Args:
        limit1 (int): Número mínimo aceptado en el juego
        limit2 (int): Número máximo aceptado en el juego
        n (int, optional): Valor a descubrir. Defaults to 10.

    Returns:
        int: Número de veces que se jugó hasta ganar
    """
    conteo = 0
    x = 0
    while x != n:
        conteo += 1
        print("Empieza el juego")
        x = int(input("Dame un número"))
        if x > 100:
            print("Tramposo")
            break
        elif x < limit1:
            print(f"Solo acepto valores mayores a {limit1}")
            continue
        elif x > limit2:
            print(f"Solo acepto valores menores a {limit2}")
            continue
        else:
            print("Ganaste!")
    return (conteo)
````

````python
def status_Juego(x, limit1, limit2):
    """Función para determinar el estatus del juego (no terminarlo)

    Args:
        x (_type_): _description_
        limit1 (_type_): _description_
        limit2 (_type_): _description_
    """
    if x > 100:
        return("Tramposo")
    elif x < limit1:
        return(f"Solo acepto valores mayores a {limit1}")
    elif x > limit2:
        return(f"Solo acepto valores menores a {limit2}")
    else:
        return("Ganaste!")
````

````python
def Adivina_n (limit1:int , limit2: int, n:int = 10) -> int:
    """Juego de adivinar un número

    Args:
        limit1 (int): Número mínimo aceptado en el juego
        limit2 (int): Número máximo aceptado en el juego
        n (int, optional): Valor a descubrir. Defaults to 10.

    Returns:
        int: Número de veces que se jugó hasta ganar
    """
    conteo = 0
    x = 0
    while x != n:
        conteo += 1
        print("Empieza el juego")
        x = int(input("Dame un número"))
        print(status_Juego(x, limit1=limit1, limit2=limit2))
    return (conteo)
````

````python
def suma_de_cuadrados(lista_numeros):
    #Primero sacamos al cuadrado cada número
    cuadrados = list()
    for n in lista_numeros:
        cuadrados.append(n**2)
    #Ahora, vamos a sumarlos
    suma_numeros = 0
    for n in cuadrados:
        suma_numeros += n
    # Regresamos la suma de cuadrados
    return(suma_numeros)
````

````python
suma_de_cuadrados([1, 2, 3, 4, 5])
````

````python
def nums_cuadrados(lista_numeros):
    """Función que sacará el cuardado de cada número en la lista

    Args:
        lista_numeros (list): Lista de números
    """
    cuadrados = list()
    for n in lista_numeros:
        cuadrados.append(n**2)
    return(cuadrados)
````

````python
def suma_numeros(lista_numeros):
    suma_numeros = 0
    for n in lista_numeros:
        suma_numeros += n
    # Regresamos la suma de cuadrados
    return(suma_numeros)
````

# Decoradores

Un decorador es una función que recibe otra función (o método, o clase) como argumento, y devuelve una nueva función modificada o extendida. Se usan principalmente para añadir funcionalidades a funciones o métodos sin modificar su código original.

````python
def saludar() -> None:
    """Función de saludo
    """
    print("Hola, mundo!")
````

````python
saludar()
````

````python
def mi_decorador(func: function) -> function:
    """Mi primer decorador

    Args:
        func (function): Función a envolver

    Returns:
        function: Función envolvida
    """
    def wrapper():
        print("Antes de llamar la función")
        saludar()
        func()
        print("Después de llamar la función")
    return wrapper

@mi_decorador
def saludar():
    print("Hola, mundo!")

saludar()
````

````python
def decorador_con_args(func):
    def wrapper(*args, **kwargs):
        print("Ejecutando:", func.__name__)
        return func(*args, **kwargs)
    return wrapper

@decorador_con_args
def sumar(a, b):
    return a + b

print(sumar(2, 3))  # Ejecutando: sumar \n 5
````

Python incluye varios decoradores útiles:

| Decorador                 | Propósito                                          |
| ------------------------- | -------------------------------------------------- |
| `@staticmethod`           | Define un método estático                          |
| `@classmethod`            | Método que recibe la clase como primer argumento   |
| `@property`               | Convierte un método en un atributo de solo lectura |
| `@functools.lru_cache`    | Crea una caché de resultados para funciones        |
| `@final` (desde `typing`) | Previene sobrescritura de métodos o clases         |
