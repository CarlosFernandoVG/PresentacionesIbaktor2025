```python
def get_weather(lat, lon, API_key):
    try:
        url = f"https://api.openweathermap.org/data/2.5/weather?lat={lat}&lon={lon}&appid={API_key}"
        response = requests.get(url)
        response.raise_for_status()
        data = response.json()
        
        return {
            "temperatura": data["main"]["temp"],
            "descripcion": data["weather"][0]["description"],
            "humedad": data["main"]["humidity"],
            "viento": data["wind"]["speed"]
        }
    except requests.exceptions.RequestException as e:
        print("Error de conexión:", e)
    except KeyError:
        print("No se pudo obtener el clima para las coordenadas:", lat, lon)
    return None
```

# Web Scrapping

+ Utiliza requests y BeautifulSoup.
+ Limpia los títulos con expresiones regulares para eliminar símbolos innecesarios.


```python
from bs4 import BeautifulSoup
import re
```

```python
def get_city_news(city):
    try:
        url = f"https://news.google.com/search?q={city}"
        r = requests.get(url)
        soup = BeautifulSoup(r.text, "html.parser")
        titles = soup.find_all("a", class_="DY5T1d", limit=3)
        news = []
        for t in titles:
            clean_title = re.sub(r'[^\w\s]', '', t.text)
            news.append(clean_title)
        return news
    except Exception as e:
        print("Error al obtener noticias:", e)
        return []
```
