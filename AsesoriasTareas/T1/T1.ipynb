```python
from functools import reduce
import csv
from typing import List, Dict
from datetime import datetime

#Lectura de archivos
def read_sales_file(filename: str) -> List[Dict]:
    datos = []
    try:
        with open(filename, 'r', newline='') as file:
            spamreader = csv.reader(file)
            #contenido = file.read()
            datos_aux = []
            for row in spamreader:
                datos_aux.append(row)
            columnas = datos_aux[0]
            renglones = datos_aux[1:]
            #Limpieza de renglones
            for r in range(len(renglones)):
                #print(int(renglones[r][3]))
                format_string = "%Y-%m-%d"
                renglones[r][0] = datetime.strptime(renglones[r][0], format_string)
                renglones[r][3] = int(renglones[r][3])
                renglones[r][4] = float(renglones[r][4])
            datos.append(columnas)
            datos.append(renglones)
    except FileNotFoundError:
        # Tu código aquí
        print("Archivo no encontrado")
    except PermissionError:
        print("Error de permiso")
    except Exception:
        print("Otro error pasó")
    return datos
```

```python
data = read_sales_file("ventas.csv")
```

```python
def process_sales_data(sales_data: List[Dict]):
    #Diccionario que va a regresar todo el procesamiento:
    dict_process_sales = dict()
    
    #Obtenemos primero todas la categorías
    categorias = list()
    for renglon in sales_data[1]:
        categorias.append(renglon[2])
    categorias = list(set(categorias))
    
    #Calculamos el total de ventas por categoría
    total_ventas_categoria = dict()
    for c in categorias:
        #print(c)
        #Filtramos para cada categoría
        datos_categoria = list(filter(lambda x: x[2] == c, sales_data[1]))
        #A cada valor de la categoría, se calcula la venta total
        ventas_totales = list(map(lambda x: x[3]*x[4], datos_categoria))
        #Todas las ventas totales se suman por cada categoría
        venta_total = reduce(lambda x, y: x+y, ventas_totales)
        #Guardo en el dic total_ventas_categoria la categoría y su venta total
        total_ventas_categoria[c] = venta_total
        
    dict_process_sales["Total de ventas por categoría"] = total_ventas_categoria
        
    #Productos con ventas superiores a un umbral
    products_ventas_superiores_5000 = dict()
    dict_process_sales["Productos con ventas superiores a 5000"] = products_ventas_superiores_5000
    
    #Calculamos el promedo de ventas por día
    prom_ventas_dia = dict()
    dict_process_sales["Promedio de ventas por día"] = prom_ventas_dia
    
    #Producto más vendido
    most_sale_prod = "Mouse"
    dict_process_sales["Producto más vendido"] = most_sale_prod
    
    return dict_process_sales
```

```python
process_data = process_sales_data(data)
```

```python
def generate_report(analysis_results: Dict, output_file: str):
    """Función generadora de reportes

    Args:
        analysis_results (Dict): Resultados....
        output_file (str): Ruta de salida del reporte...
    """
    #Creamos o remplazamos el archivo
    file_1 = open(output_file, 'w') 
    #Escribimos el total de ventas por categoría
    file_1.write("El total de ventas por categoría queda desglozado en la siguiente lista:\n")
    file_1.write("\n")
    for dato in analysis_results["Total de ventas por categoría"].items():
        file_1.write(f"En la categoría {dato[0]}, se obtuvo una venta total de {dato[1]}\n")
    #Escribimos el producto más vendido
    #Cerramos el archivo
    file_1.close()
```
