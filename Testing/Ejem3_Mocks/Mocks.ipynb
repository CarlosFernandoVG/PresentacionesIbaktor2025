Un mock es una versión falsa o simulada de un objeto, función o servicio externo. Se utiliza para:

Aislar la unidad que se quiere probar.
Evitar llamadas reales (por ejemplo, a bases de datos, APIs, servicios externos, etc.).
Controlar lo que devuelve un objeto o función.

Para ver más del uso de Mocks véase:

https://docs.python.org/3/library/unittest.mock-examples.html

Ejemplo

```python
import unittest
from unittest.mock import Mock

# Función que queremos probar
def saluda(api_usuario):
    nombre = api_usuario.obtener_nombre()
    return f"Hola, {nombre}"

# Prueba con mock
def test_saluda():
    mock_api = Mock()
    mock_api.obtener_nombre.return_value = "Carlos"
    resultado = saluda(mock_api)
    assert resultado == "Hola, Carlos"
    
unittest.main(argv=[''], verbosity=2, exit=False)
#argv=[''] evita que unittest tome argumentos del Jupyter.
#exit=False evita que el kernel se detenga después de correr.
```

```python
##Si guardamos el test anterior en el archivo test_algo.py se puede ejecutar lo siguiente
#!python3 -m unittest test_algo.py
```

Aquí:

+ `mock_api` simula un objeto que tiene la función `obtener_nombre()`.
+ Evitamos depender de una API real, pero controlamos lo que devuelve.

El módulo `unittest.mock` permite simular (mockear) objetos, funciones, clases, métodos, etc., para aislar el comportamiento de lo que quieres probar. Hay varios métodos a tener en cuenta

1. `return_value`

Define qué debe devolver el mock cuando se llama como función o método.

```python
from unittest.mock import Mock

mock_func = Mock()
mock_func.return_value = 42

assert mock_func() == 42
```

2. `side_effect`

Permite:

+ Lanzar una excepción
+ Devolver múltiples valores
+ Usar una función personalizada

```python
# Como función secuencial
mock = Mock()
mock.side_effect = [1, 2, 3]
print(mock())  # 1
print(mock())  # 2

# Como excepción
mock_error = Mock()
mock_error.side_effect = ValueError("¡Error simulado!")
# mock_error()  # lanza ValueError

# Como función
def cuadrado(x):
    return x * x

mock_funcional = Mock()
mock_funcional.side_effect = cuadrado
print(mock_funcional(4))  # 16
```

3. `assert_called()` y `assert_called_once()``

Verifican que el mock fue llamado.

```python
mock = Mock()
mock()

mock.assert_called()         # ✅
mock.assert_called_once()    # ✅
```

4. `assert_called_with(*args, **kwargs)`

Verifica que fue llamado con ciertos argumentos.

```python
mock = Mock()
mock("Carlos", edad=30)

mock.assert_called_with("Carlos", edad=30)
```

5. `assert_called_once_with(*args, **kwargs)`

Igual que el anterior, pero asegura que fue llamado una sola vez con esos valores.

6. `assert_not_called()`

Verifica que nunca fue llamado.

```python
mock = Mock()
mock.assert_not_called()  # ✅
```

7. `call_args` y `call_count`

+ `call_args`: devuelve los últimos args usados.
+ `call_count`: cuántas veces fue llamado.

```python
mock = Mock()
mock(1, 2)
mock("a", b="B")

print(mock.call_count)     # 2
print(mock.call_args)      # ('a',), {'b': 'B'}
```

8. `reset_mock()`

Reinicia el historial del mock.

```python
mock = Mock()
mock(1)
mock.reset_mock()
mock.assert_not_called()
```

9. `mock_calls`

Devuelve una lista de todas las llamadas con sus args:

```python
mock = Mock()
mock(1)
mock(2, 3)

print(mock.mock_calls)
# [call(1), call(2, 3)]
```
