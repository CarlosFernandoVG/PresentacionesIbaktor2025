```python
query = """
SELECT 
    mes,
    ventas_totales,
    num_productos,
    ventas_totales / num_productos as venta_promedio_por_producto
FROM (
    SELECT 
        DATE_FORMAT(fecha, 'Año: %Y, mes: %M') as mes,
        SUM(cantidad * precio) as ventas_totales,
        COUNT(DISTINCT producto) as num_productos
    FROM ventas
    GROUP BY DATE_FORMAT(fecha, 'Año: %Y, mes: %M')
) AS metricas_mensuales
ORDER BY mes;
"""
```

```python
pd.read_sql(query, engine)
```

```python
df_aux = pd.read_sql(query, engine)
```

```python
import mariadb

# Conecta a la base de datos
try:
    conn = mariadb.connect(
        user="python_user",
        password="",
        host="localhost",
        port=3306,
        database="IbaktorExamples"
    )
    print("Conexión exitosa a MariaDB")

except mariadb.Error as e:
    print(f"Error al conectar a MariaDB: {e}")
    exit(1)
```

```python
# Ejecutar una consulta
cur.execute("SELECT NOW();")
```
