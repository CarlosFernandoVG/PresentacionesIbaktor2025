```python
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
housing = fetch_california_housing(as_frame=True)
df = housing.frame

X = df.drop("MedHouseVal", axis=1)
y = df["MedHouseVal"]

# Escalado
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# División train/test
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
rf = RandomForestRegressor(random_state=42)

param_grid = {
    "n_estimators": [50, 100, 200],    # número de árboles
    "max_depth": [None, 10, 20],       # profundidad máxima
    "min_samples_split": [2, 5, 10],   # división mínima
    "min_samples_leaf": [1, 2, 4],     # mínimo en hoja
    "bootstrap": [True, False]         # muestreo bootstrap
}
```

```python
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    n_jobs=-1,
    verbose=2,
    scoring="neg_mean_squared_error"   # métrica para regresión
)

grid_search.fit(X_train, y_train)

print("Mejores hiperparámetros encontrados:")
print(grid_search.best_params_)
```

```python
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Evaluación del mejor modelo (Random Forest optimizado):")
print("RMSE:", rmse)
print("R²:", r2)
```
