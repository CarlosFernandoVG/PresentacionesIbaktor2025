Faker es una librería de Python que sirve para generar datos falsos (dummy data) de manera sencilla. Muy útil para crear conjuntos de datos para probar aplicaciones, hacer demostraciones o entrenar modelos con datos simulados. Algunos de los datos que se pueden simular con Faker son:

+ Persona
    * fake.name() → Nombre completo
    * fake.first_name() → Nombre
    * fake.last_name() → Apellido
    * fake.prefix() → Prefijo (Sr., Dr., etc.)
    * fake.suffix() → Sufijo (Jr., PhD, etc.)
    * fake.simple_profile() → Perfil simple (nombre, sexo, etc.)
    * fake.profile() → Perfil más completo
+ Dirección
    * fake.address() → Dirección completa
    * fake.street_name() → Nombre de la calle
    * fake.street_address() → Dirección de calle
    * fake.city() → Ciudad
    * fake.state() → Estado/provincia
    * fake.country() → País
    * fake.postcode() → Código postal
    * fake.geo_coordinate() → Coordenada geográfica
+ Contacto / Internet
    * fake.email() → Correo electrónico
    * fake.safe_email() → Correo seguro (ej: ejemplo@example.com)
    * fake.user_name() → Nombre de usuario
    * fake.password() → Contraseña
    * fake.url() → URL
    * fake.domain_name() → Dominio
    * fake.ipv4() / fake.ipv6() → IPs
+ Teléfonos
    * fake.phone_number() → Número de teléfono
    * fake.msisdn() → Número de teléfono móvil
+ Fechas / Tiempo
    * fake.date() → Fecha
    * fake.date_time() → Fecha y hora
    * fake.date_of_birth() → Fecha de nacimiento
    * fake.future_date() → Fecha futura
    * fake.past_date() → Fecha pasada
+ Finanzas
    * fake.credit_card_number() → Número de tarjeta de crédito
    * fake.credit_card_provider() → Tipo de tarjeta (Visa, MasterCard, etc.)
    * fake.credit_card_expire() → Fecha de expiración
+ Texto
    * fake.word() → Una palabra
    * fake.sentence() → Una oración
    * fake.paragraph() → Un párrafo
    * fake.text() → Texto largo
+ Trabajo / Empresa
    * fake.company() → Nombre de empresa
    * fake.job() → Profesión
    * fake.catch_phrase() → Eslogan
+ Otros
    * fake.random_int() → Entero aleatorio
    * fake.random_number() → Número aleatorio
    * fake.boolean() → Verdadero/Falso
    * fake.color_name() → Nombre de color
    * fake.language_name() → Nombre de idioma
    * fake.currency_name() → Nombre de moneda
    * fake.license_plate() → Placa de vehículo

```python
from faker import Faker

fake = Faker("es_MX")
Faker.seed(1234)

print(fake.name())         
print(fake.address())      
print(fake.email())        
print(fake.date_of_birth())
print(fake.date_of_birth())
print(fake.company())
print(fake.date())
```

```python
Faker.seed(1234)
random.seed(1234)
np.random.seed(1234) 

# Dataset ficticio de ventas
data = {
    "fecha": [fake.date_between(start_date=start, end_date=end) for _ in range(100)],
    "producto": [random.choice(productos) for _ in range(100)],
    "cantidad": np.random.randint(1, 101, size=100),
    "precio_unitario": np.random.randint(500, 10000, size=100),
    "ciudad": random.choices(["Ciudad de México", "Guadalajara", "Monterrey", "Puebla", "Tijuana"], k=100),
}

df = pd.DataFrame(data)
```

```python
# Calcular ingresos
df["ingreso"] = df["cantidad"] * df["precio_unitario"]
```

```python
coords = {
    "Ciudad de México": [19.4326, -99.1332],
    "Guadalajara": [20.6597, -103.3496],
    "Monterrey": [25.6866, -100.3161],
    "Puebla": [19.0413, -98.2062],
    "Tijuana": [32.5149, -117.0382]
}

df["lat"] = df["ciudad"].map(lambda c: coords[c][0])
df["lon"] = df["ciudad"].map(lambda c: coords[c][1])
```
